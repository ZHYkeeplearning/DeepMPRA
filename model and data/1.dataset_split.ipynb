```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
file_path = r"..."
df = pd.read_csv(file_path)
```

```python
unique_lakes = df['Lake'].unique()
train_lakes, test_lakes = train_test_split(
    unique_lakes,
    test_size=0.2,
    random_state=42,
    shuffle=True
)
train_df = df[df['Lake'].isin(train_lakes)].copy()
test_df = df[df['Lake'].isin(test_lakes)].copy()
output_dir = r"..."
train_df.to_csv(f"{output_dir}\\train.csv", index=False)
test_df.to_csv(f"{output_dir}\\test.csv", index=False)
```
